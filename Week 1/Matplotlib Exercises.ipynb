### Read Total Profit of all months and show it using a line plot

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
data = pd.read_csv("company_sales_data.csv")

# Extract data
months = data['month_number']
profits = data[' total_profit']

# Create line plot
plt.plot(months, profits, marker='o')

# Add labels and title
plt.title('Total Profit Per Month')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')

# Show grid
plt.grid(True)

# Display plot
plt.show()
```

### Read data for Bathing Soap and Facewash and display using Subplots

```python
# Extract data
months = data['month_number']
bathingsoap = data[' bathingsoap']
facewash = data[' facewash']

# Create subplot figure
fig, axes = plt.subplots(2, 1, figsize=(8, 6))

# Bathing soap subplot
axes[0].plot(months, bathingsoap, marker='o')
axes[0].set_title('Bathing Soap Sales')
axes[0].set_xlabel('Month')
axes[0].set_ylabel('Sales')
axes[0].grid(True)

# Facewash subplot
axes[1].plot(months, facewash, marker='o')
axes[1].set_title('Facewash Sales')
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Sales')
axes[1].grid(True)

# Adjust spacing
plt.tight_layout()

# Display plots
plt.show()
```
