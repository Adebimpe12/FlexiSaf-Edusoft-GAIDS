```python
import random

# Define the game board
board = [" " for _ in range(9)]


# Display the board
def print_board():
    print("\n")
    for i in range(0, 9, 3):
        print(board[i] + " | " + board[i + 1] + " | " + board[i + 2])
        if i < 6:
            print("--+---+--")
    print("\n")


# Check if a player has won
def check_winner(player):
    winning_combinations = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        [0, 4, 8],
        [2, 4, 6]
    ]

    for combo in winning_combinations:
        if all(board[pos] == player for pos in combo):
            return True

    return False


# Check if the game is a tie
def check_tie():
    return " " not in board


# Get available moves
def available_moves():
    return [i for i in range(9) if board[i] == " "]


# AI move (simple learning strategy)
def ai_move():
    # Try to win
    for move in available_moves():
        board[move] = "O"
        if check_winner("O"):
            return
        board[move] = " "

    # Block player from winning
    for move in available_moves():
        board[move] = "X"
        if check_winner("X"):
            board[move] = "O"
            return
        board[move] = " "

    # Otherwise choose random move
    move = random.choice(available_moves())
    board[move] = "O"


# Human move
def human_move():
    while True:
        try:
            move = int(input("Enter your move (1-9): ")) - 1

            if move in available_moves():
                board[move] = "X"
                break
            else:
                print("Invalid move. Try again.")

        except ValueError:
            print("Please enter a number between 1 and 9.")


# Main game loop
def play_game():
    print("Welcome to Tic-Tac-Toe!")
    print("You are X and AI is O")

    while True:
        print_board()

        # Human turn
        human_move()

        if check_winner("X"):
            print_board()
            print("You win!")
            break

        if check_tie():
            print_board()
            print("It's a tie!")
            break

        # AI turn
        ai_move()

        if check_winner("O"):
            print_board()
            print("AI wins!")
            break

        if check_tie():
            print_board()
            print("It's a tie!")
            break


# Call the main game loop
play_game()
```
