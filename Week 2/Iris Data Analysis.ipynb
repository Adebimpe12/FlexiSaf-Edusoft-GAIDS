# Iris dataset

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

# Load dataset
iris = load_iris()

# Convert to DataFrame
data = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add target column
data['species'] = iris.target

# Map target labels to species names
data['species'] = data['species'].map({
    0: 'setosa',
    1: 'versicolor',
    2: 'virginica'
})
```

```python
print("First 5 rows:")
print(data.head())
```

### Check data types

```python
print("\nData Information:")
print(data.info())
```

### Summary statistics

```python
print("\nSummary Statistics:")
print(data.describe())
```

### Mean

```python
print("\nMean:")
print(data.mean(numeric_only=True))
```

### Median

```python
print("\nMedian:")
print(data.median(numeric_only=True))
```

### Mode

```python
print("\nMode:")
print(data.mode(numeric_only=True).iloc[0])
```

### Standard Deviation

```python
print("\nStandard Deviation:")
print(data.std(numeric_only=True))
```

### Basic Statistics

```python
stats = pd.DataFrame({
    'Mean': data.mean(numeric_only=True),
    'Median': data.median(numeric_only=True),
    'Mode': data.mode(numeric_only=True).iloc[0],
    'Std Dev': data.std(numeric_only=True)
})

print("\nBasic Statistics:")
print(stats)
```

### Correlation

```python
correlation = data.corr(numeric_only=True)

print("\nCorrelation Matrix:")
print(correlation)
```
